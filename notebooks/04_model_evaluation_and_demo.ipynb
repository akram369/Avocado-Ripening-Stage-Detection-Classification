## 3. Mandatory End-of-Assignment Questions & Answers

### Q1: In production, how does your model perform?
- **Inference Speed:** Classical SVM executes feature extraction + classification in **~0.31 ms / image**, while MobileNetV2 achieves **~15.95 ms / image on CPU** (and <5 ms on edge GPUs like NVIDIA Jetson Nano/Orin). Both comfortably surpass the typical 30-50 FPS sorting line requirement (20-33 ms budget).
- **Environmental Shifts:** Lighting glare, specular reflections on waxy fruit peel, and conveyor shadows can shift hue/saturation features. The MobileNetV2 model demonstrates superior robustness to varying lighting and scale due to spatial receptive fields and multi-layer feature representations.

### Q2: What are the drawbacks of your model?
- **Color-Only / Visible Surface Ambiguity:** External color and skin texture do not always correlate perfectly with internal pulp firmness or core breakdown.
- **Single-Perspective Blind Spots:** A single overhead camera cannot detect stem-end rot or bruising on the fruit undersides.
- **Dataset Bias:** The `fresh` class had resolution differences (small PNGs) vs diseased classes (800x800 JPGs), requiring strict resizing and standardization to avoid shortcut learning.

### Q3: If given more time, what would you improve?
- **Multi-Angle / Rotating Conveyor Imaging:** Use multi-camera inspection with rolling conveyor shafts for 360-degree surface coverage.
- **Multimodal Sensor Fusion:** Pair visible RGB vision with Non-Destructive NIR (Near-Infrared) spectroscopy or acoustic firmness sensors.
- **Model Optimization:** Export PyTorch models to ONNX and INT8 TensorRT for 3-5x edge acceleration.

### Q4: Where did you spend the most time while solving your assignment?
- **~40% on Feature Engineering & Color Space Tuning:** Designing handcrafted color histograms (CIELAB a*, b*, HSV H, S) and multiscale LBP + GLCM texture descriptors.
- **~30% on Data Pipeline & Bias Mitigation:** Standardizing PNG transparency, correcting resolution shifts, and addressing class imbalance via balanced weighting.
- **~30% on Model Benchmarking & Metric Evaluation:** Training both approaches, fine-tuning hyperparameters, and building rigorous comparative tables.
